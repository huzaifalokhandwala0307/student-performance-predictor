```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report

# Load iris dataset
iris = load_iris()
x = iris.data
y = iris.target

# Splitting data
Xa, Xb, Ya, Yb = train_test_split(x, y, test_size=0.6, random_state=26)

# KNN model
knn = KNeighborsClassifier(n_neighbors=9)
knn.fit(Xa, Ya)

# Prediction
y_pred = knn.predict(Xb)

# 🔹 Evaluation Metrics
accuracy  = accuracy_score(Yb, y_pred)
precision = precision_score(Yb, y_pred, average='macro')   # macro = equal weight to each class
recall    = recall_score(Yb, y_pred, average='macro')
f1        = f1_score(Yb, y_pred, average='macro')

print("\n----- Practical 7 Evaluation Metrics -----")
print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1 Score  : {f1:.4f}")

print("\nClassification Report:\n")
print(classification_report(Yb, y_pred, target_names=iris.target_names))
```
